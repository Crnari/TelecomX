## Bitácora de decisiones — Fase 3

### Decisión 1 — Los 224 registros sin churn

**Qué encontré.** 224 registros (3,08% del total) tienen la columna `churn` vacía. No están como nulos, están como texto vacío `''`. Por eso `isna()` e `info()` no los detectan: para pandas, un texto vacío no es un nulo.

**Qué opciones tenía.** Dejarlos en todo el análisis. Dejarlos solo para describir la cartera y sacarlos de los cruces contra churn. O sacarlos del análisis.

**Qué elegí.** Sacarlos. El análisis queda con 7.043 registros, y ese es el número base de todo el proyecto.

**Por qué.** El churn es la variable que quiero explicar. Si le invento un valor a un registro que no lo tiene, estoy inventando el resultado. Un target no se imputa.

Descarté la opción intermedia porque me dejaba dos números distintos dando vueltas por el mismo informe, y es fácil confundirse al leerlo.

Antes de sacarlos revisé que no fueran un grupo raro. Comparé los 224 contra los 7.043 en contrato, método de pago, tipo de internet y antigüedad. La diferencia más grande fue de 4,46 puntos porcentuales (DSL), el resto por debajo de 3. La mediana de antigüedad da igual en los dos grupos: 29 meses. O sea, se reparten como el resto de la cartera y sacarlos no deforma la muestra.

Igual no desaparecen del proyecto: que falte el churn en el 3% de los registros es un problema de calidad de datos y va al informe.

**Qué asumo.** Que el churn falta por un fallo al registrar el dato, y no por algo que tenga que ver con el cliente. La comparación apoya esto pero no lo prueba: solo revisé cuatro variables. Si esos 224 fueran, por ejemplo, clientes cuyo registro se rompió justo cuando se estaban dando de baja, mi tasa de churn saldría más baja de lo real. Con los datos que tengo no lo puedo descartar.

---

### Decisión 2 — Los 11 registros sin `total_charges`

**Qué encontré.** 11 registros (0,15%) tienen `total_charges` vacío. Los 11 tienen `tenure = 0` y todos sí tienen `monthly_charges`. Por culpa de esos vacíos, la columna entera figuraba como texto. Ninguno de los 11 estaba entre los 224 que saqué en la decisión anterior.

**Qué opciones tenía.** Dejarlos como nulo. Ponerles 0. O borrar los registros.

**Qué elegí.** Ponerles 0 y convertir la columna a número.

**Por qué.** No estoy imputando, estoy deduciendo. Imputar es estimar un valor cuando no sabes cuál es. Acá sí lo sé: si el cliente lleva 0 meses, no ha acumulado nada. El valor sale del propio dato.

Poner nulo sería decir "no sé cuánto pagó", y sí lo sé. Además tiene un efecto práctico: la columna se quedaría como texto y esos 11 clientes desaparecerían sin avisar de cualquier promedio o gráfico de `total_charges` más adelante.

Borrarlos era lo más destructivo para un valor que puedo deducir.

**Nota al margen (no es un hallazgo).** De los 11, hay 10 con contrato de dos años y 1 con contrato de un año. Con 11 registros no se puede concluir nada, así que esto no va al informe. Lo dejo anotado nada más: son clientes que acaban de firmar el contrato más largo que ofrece la empresa.

**Qué asumo.** Que `tenure` está bien registrado, y que `tenure = 0` significa cliente nuevo al que todavía no le han facturado, no un error en esa columna.

**Qué cambia para después.** `total_charges` ahora tiene mínimo 0. Esos 11 van a aparecer en el extremo bajo de cualquier gráfico de esa variable.

---

### Decisión 3 — `senior_citizen` en 0/1

**Qué encontré.** `senior_citizen` viene como número entero (0 y 1), mientras que las demás columnas binarias del dataset usan `Yes`/`No`.

**Qué opciones tenía.** Dejarla como viene, o pasarla a `Yes`/`No`.

**Qué elegí.** Pasarla a texto: `1 → Yes`, `0 → No`. Antes revisé que la columna solo tuviera esos dos valores.

**Por qué.** Al ser número, aparece en `describe()` al lado de `tenure`, `monthly_charges` y `total_charges`, y su media se lee como si fuera un promedio. Pero no es un promedio, es una proporción. Eso confunde a quien lea el resumen.

Esto va un poco en contra de lo que decidí en la Fase 1, cuando dejé los nombres de columna en inglés para no alejarme de la fuente. Sigo con ese criterio, pero con un límite: si la fuente clasifica mal la variable, ahí sí la corrijo. `senior_citizen` es categórica, no numérica.

**Qué asumo.** Que 0 es No y 1 es Yes, que es lo habitual. No hay documentación de la fuente que lo diga.

---

### Decisión 4 — Variable nueva: `service_count`

**Para qué la creo.** Para responder esto: ¿los clientes que contratan más servicios adicionales se quedan más tiempo? Ninguna de las seis columnas de servicios responde eso por separado. Cada una me dice si tiene ese servicio puntual, pero no qué tan enganchado está el cliente con la empresa.

**Cómo la construí.** Cuento cuántos `Yes` tiene cada cliente en seis columnas: `online_security`, `online_backup`, `device_protection`, `tech_support`, `streaming_tv` y `streaming_movies`. Da un número de 0 a 6.

**Por qué esas seis y no otras.** Son las que el cliente puede elegir contratar o no. `phone_service` e `internet_service` quedan fuera porque son el servicio base: casi todos tienen alguno, así que no sirven para diferenciar. `multiple_lines` también es un adicional, pero la dejé fuera para que las seis midan lo mismo: todas dependen de tener internet contratado, mientras que `multiple_lines` depende de tener teléfono.

**Qué cambié antes de contar.** En las seis columnas convertí `No internet service` en `No`. Si el cliente no tiene internet, no puede tener el adicional, así que en la práctica es lo mismo que no tenerlo. Este cambio pisa las columnas originales y no se puede deshacer, pero no pierdo información: quién no tiene internet se sabe mirando `internet_service`.

**Cómo verifiqué que quedó bien.** Los valores van de 0 a 6, ninguno se sale. Y el control importante: todos los clientes sin internet quedaron con `service_count = 0`.

**Un problema que ya detecté y que afecta a la Fase 5.** El grupo de `service_count = 0` tiene 2.219 clientes, pero no son todos iguales. 1.526 no tienen internet, o sea que nunca pudieron contratar adicionales. Los otros 693 sí tienen internet y decidieron no contratar ninguno. Es el mismo número con dos significados opuestos.

Por eso, cuando cruce `service_count` contra churn, voy a usar solo los clientes que tienen internet: 5.517 registros, con el grupo 0 reducido a 693. El número lo voy a dejar escrito en el gráfico para que se vea que ahí el denominador cambia.

Los 1.526 sin internet siguen en todos los demás cruces. Y su comportamiento lo miro aparte, cruzando `internet_service` contra churn.

## Hipótesis previas al análisis

*Escritas el 2026-08-04, antes de ejecutar el EDA. No se modifican después.*

**1. — Tipo de contrato.**
Los clientes con contrato mes a mes abandonan más que los de contrato largo.
Mecanismo: el contrato mes a mes permite cancelar en cualquier momento, mientras
que los contratos largos obligan a esperar hasta el vencimiento. La retención del
contrato largo sería una barrera de salida, no necesariamente mayor satisfacción.

**2. — Método de pago.**
Los clientes con métodos de pago manuales abandonan más que los de pago automático.
Mecanismo: cada cobro manual es una decisión activa del cliente y una oportunidad
de no pagar. El pago automático sigue corriendo por inercia.

**3. — Antigüedad.**
Los clientes con menos de 12 meses de permanencia abandonan más que los de mayor
antigüedad.
Mecanismo: insatisfacción temprana con el servicio o aparición de mejores ofertas
de la competencia, sumado a menor fricción de salida en los primeros meses.
Nota: el umbral de 12 meses es una elección propia para hacer la hipótesis
contrastable, no un corte derivado del negocio. En la Fase 5 se verifica si el
quiebre real está en ese punto o en otro.

## Lista de diagnóstico

*Fase 2 — Auditoría previa a cualquier corrección. Ningún dato ha sido modificado.*

**1. Vacíos en el target**

La columna `churn` presenta 224 registros con cadena vacía `''`, equivalente al 3,08%
de los 7.267 registros. Estos valores no son detectados por `isna()` ni por `info()`:
pandas solo considera nulos a `NaN`, `None` y `NaT`, mientras que `''` es una cadena
de texto válida de longitud cero. Fue necesaria una comprobación explícita por
comparación directa para encontrarlos.

**2. `total_charges` no numérica**

La columna llega con dtype `object` en lugar de `float64`. La causa son 11 registros
(0,15%) que quedan vacíos al eliminar espacios, lo que impide la conversión a
numérico: basta un valor no convertible para degradar el tipo de toda la columna.
Los 11 registros tienen `tenure = 0`, por lo que la ausencia es estructuralmente
coherente — sin meses de permanencia no hay cargo acumulado que registrar.
Contraste que lo confirma: `monthly_charges`, del mismo bloque y con el mismo
formato de origen, sí convirtió a `float64`.

**3. Inconsistencia de codificación**

`senior_citizen` viene como `int64` (0/1), mientras que el resto de variables
binarias del dataset usan texto (`Yes`/`No`).

**4. Duplicados — verificado, sin hallazgos**

`customer_id` no presenta ningún valor repetido. Al ser el identificador único por
cliente, su unicidad hace imposible la existencia de filas idénticas en las 21
columnas, por lo que ambos casos quedan descartados con una sola verificación.

**5. Relación no aritmética entre cargos** *(nota de contexto, no es un error)*

`monthly_charges × tenure` no reproduce el valor de `total_charges`. Las variables
sí están relacionadas —el acumulado depende del cargo y de la antigüedad— pero no
existe equivalencia aritmética exacta: `monthly_charges` es la tarifa vigente hoy,
mientras que `total_charges` es el acumulado histórico. Cambios de plan,
promociones o ajustes de tarifa rompen la identidad. Se documenta como
característica del dataset, no como inconsistencia a corregir.

**Verificación adicional:** las 18 columnas de texto fueron revisadas con
`.str.strip()`. Únicamente `churn` y `total_charges` presentan valores vacíos.
